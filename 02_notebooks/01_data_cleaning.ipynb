# 2. Data Cleaning
The purpose of this stage is to improve the quality and consistency of the Olist e-commerce datasets before performing sales analysis and visualization.

Based on the initial data inspection, the following data-cleaning tasks will be performed:
- **Handling Missing Values**: Investigate missing values and determine appropriate treatments without removing useful records unnecessarily.
- **Data Type Conversion**: Convert date and timestamp columns to appropriate datetime formats and verify other column data types.
- **Duplicate Removal**: Identify and remove unnecessary duplicate records while preserving valid data.
- **Data Validation**: Verify the cleaned datasets for consistency, accuracy, and completeness.

## 2.1 Handling Missing Values
Missing values were identified during the initial data inspection in the orders, order_reviews, and products datasets.
Before handling these values, their patterns and possible causes will be investigated. Depending on the purpose of each column, missing values may be retained, filled with an appropriate placeholder, or handled through another justified method.

The objective is to preserve useful information while improving data quality and maintaining the reliability of subsequent analysis.

```python
# Check missing values in all nine datasets

for name, df in datasets.items():
    missing_count = df.isnull().sum()
    missing_count = missing_count[missing_count > 0]

    if not missing_count.empty:
        print(f"\n{name}")
        print(missing_count.sort_values(ascending=False))
    else:
        print(f"\n{name}: No missing values")
```

```python
# Investigating missing timestamps by order status

timestamp_columns = [
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date"
]

for col in timestamp_columns:
    print(f"\nMissing values in {col}:")

    print(
        orders.loc[orders[col].isna(), "order_status"]
        .value_counts(dropna=False)
    )
```

```python
# Investigating delivered orders with missing timestamps

delivered_orders = orders[orders["order_status"] =="delivered"]

print("Delivered orders with missing approval timestamps:")
display(
    delivered_orders[
        delivered_orders["order_approved_at"].isna()
    ]
)

print("Delivered orders with missing carrier timestamps:")
display(
    delivered_orders[
        delivered_orders["order_delivered_carrier_date"].isna()
    ]
)

print("Delivered orders with missing customer delivery timestamps:")
display(
    delivered_orders[
        delivered_orders["order_delivered_customer_date"].isna()
    ]
)
```

### Investigation of Missing Order Timestamps

The missing timestamps were investigated by comparing them with the corresponding order statuses.

Fourteen delivered orders have missing approval timestamps, two delivered orders have missing carrier timestamps, and eight delivered orders have missing customer delivery timestamps.

These records indicate that some orders have incomplete timestamps even though their status is marked as delivered. Since the actual missing dates cannot be reliably inferred from the available information, they will not be replaced with estimated or arbitrary values.

**Cleaning decision:** Preserve the affected records and retain their missing timestamps. These exceptions will be documented as data-quality limitations. No orders will be deleted solely because their timestamps are missing.


```python
# Investigating missing product information

data_columns = [
    "product_category_name",
    "product_name_lenght",
    "product_description_lenght",
    "product_photos_qty"
]

print("Missing values per data column:")
print(products[data_columns].isna().sum())

print("\nRows where all four data columns are missing:")
print(products[data_columns].isna().all(axis=1).sum())

print("\nRows where at least one data column is missing:")
print(products[data_columns].isna().any(axis=1).sum())
```

### Investigation of Missing Product Metadata

The `products` dataset contains 610 missing values in each of the following columns: `product_category_name`, `product_name_lenght`, `product_description_lenght`, and `product_photos_qty`.

Further investigation confirmed that all four columns are missing together in exactly the same 610 product records. No additional records have only a subset of these four fields missing.

**Cleaning decision:** Retain these product records and preserve the missing metadata. The affected records will not be deleted solely because these fields are unavailable. Missing product categories may limit category-based analysis, while missing metadata may restrict analyses involving product descriptions, names, or photographs.


```python
# Identifying products with any missing physical measurements
measurement_columns = [
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]

missing_measurements = products[
    products[measurement_columns].isna().any(axis=1)
]

print("Number of products with missing measurements:",
      len(missing_measurements))

print("\nMissing measurements per column:")
print(missing_measurements[measurement_columns].isna().sum())

print("\nAffected product records:")
display(missing_measurements)
```

### Investigation of Missing Product Measurements

The `products` dataset contains two records with missing values in all four physical measurement columns: `product_weight_g`, `product_length_cm`, `product_height_cm`, and `product_width_cm`.

One affected product has its category and other product metadata available, while the second product also has missing category and metadata information.

The available data does not provide enough information to reliably reconstruct the missing physical measurements.

**Cleaning decision:** Retain both product records and preserve their missing measurements. No arbitrary values will be imputed, and neither record will be deleted solely because its measurements are unavailable.


```python
# Checking missing values across all datasets
for name, df in datasets.items():
    missing = df.isna().sum()
    missing = missing[missing > 0]

    print(f"\n--- {name} ---")

    if missing.empty:
        print("No missing values.")
    else:
        print(missing.sort_values(ascending=False))
```

### Summary of Missing-Value Handling

Missing values were investigated across all nine datasets. The investigation focused on understanding the affected columns and determining whether the missing information could be reliably recovered.

* **Orders:** Missing approval and delivery timestamps were investigated in relation to order status. Unavailable timestamps were preserved because their actual values could not be reliably inferred.
* **Products:** Four metadata columns were missing together in 610 product records. Two additional product records had missing values in all four physical measurement columns. These records and their missing values were retained.
* **Order reviews:** Missing review titles and comments were preserved because written feedback is optional and the remaining review information may still be useful.

No records were deleted, and no arbitrary values were introduced to replace missing information.

**Outcome:** The missing-value patterns have been investigated and documented. Missing values are retained where the available data does not support reliable imputation.

## 2.2 Data Type Conversion

Correct data types are important for accurate analysis. For example, converting timestamp columns to datetime format enables us to calculate delivery durations, analyze order trends over time, and compare actual delivery dates with estimated delivery dates.

In this section, the existing data types will be inspected before converting timestamp columns to datetime format.


```python
# Converting order timestamp columns to datetime
order_date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for column in order_date_columns:
    orders[column] = pd.to_datetime(
        orders[column],
        errors="coerce"
    )

print("Updated data types:")
print(orders[order_date_columns].dtypes)
```

```python
# Check missing timestamps after conversion
print("Missing timestamps after conversion:")
print(orders[order_date_columns].isna().sum())
```

### Datetime Conversion: Orders Dataset

The timestamp columns in the `orders` dataset were converted from string (`str`) to datetime format using `pd.to_datetime()`.

The following columns were converted:

* `order_purchase_timestamp`
* `order_approved_at`
* `order_delivered_carrier_date`
* `order_delivered_customer_date`
* `order_estimated_delivery_date`

The `errors="coerce"` parameter was used to convert valid timestamp values to datetime format while representing unparseable values as `NaT` (Not a Time).

**Verification:** The data types and missing-value counts were checked after conversion to confirm that the timestamps were converted correctly and to identify any unexpected missing values.

**Missing-value handling:** Existing missing timestamps were preserved as `NaT`. No arbitrary dates were assigned, because the actual missing values could not be reliably inferred from the available data.

**Outcome:** The timestamp columns are now suitable for time-based analysis, including order trends, delivery-duration calculations, and comparisons between estimated and actual delivery dates.

## 2.3 Handling Duplicate Records

Duplicate records can affect the accuracy of data analysis by causing certain observations to be counted more than once.

In this section, all nine datasets will be checked for exact duplicate rows. The results will be examined before any records are removed to ensure that potentially meaningful information is not lost.


```python
# Checking exact duplicate rows in each dataset

for name, df in datasets.items():
    duplicate_count = df.duplicated().sum()

    print(f"\nDataset: {name}")
    print(f"Number of duplicate rows: {duplicate_count}")
```

```python
# Display the first five rows
display(geolocation.head())

# Display the dataset dimensions
print("Dataset shape:", geolocation.shape)

# Display column names
print("\nColumn names:")
print(geolocation.columns.tolist())
```

```python
# Inspect a sample of exact duplicate rows
duplicate_rows = geolocation[
    geolocation.duplicated(keep=False)
]

print("Rows involved in exact duplication:", len(duplicate_rows))

display(duplicate_rows.head(10))
```

```python
# Calculate the number and percentage of exact duplicate rows
total_rows = len(geolocation)
duplicate_count = geolocation.duplicated().sum()
duplicate_percentage = (duplicate_count / total_rows) * 100

print("Total rows:", total_rows)
print("Exact duplicate rows:", duplicate_count)
print(f"Duplicate percentage: {duplicate_percentage:.2f}%")
```

```python
# Count unique coordinate combinations for each ZIP prefix
coordinates_per_zip = (
    geolocation
    .groupby("geolocation_zip_code_prefix")[
        ["geolocation_lat", "geolocation_lng"]
    ]
    .apply(lambda group: group.drop_duplicates().shape[0])
)

print("ZIP prefixes with multiple coordinate combinations:",
      (coordinates_per_zip > 1).sum())

print("\nMaximum unique coordinate combinations for one ZIP prefix:",
      coordinates_per_zip.max())
```

```python
# Remove exact duplicate rows from a copy of geolocation
geolocation_cleaned = geolocation.drop_duplicates().copy()

# Verify the results
print("Original shape:", geolocation.shape)
print("Cleaned shape:", geolocation_cleaned.shape)

print(
    "Rows removed:",
    len(geolocation) - len(geolocation_cleaned)
)

print(
    "Remaining exact duplicates:",
    geolocation_cleaned.duplicated().sum()
)
```

```python
# Update the dictionary to use the cleaned geolocation dataset
datasets["geolocation"] = geolocation_cleaned
```

### Duplicate Handling: Geolocation Dataset

The `geolocation` dataset initially contained 1,000,163 rows. An exact duplicate check identified 261,831 duplicate occurrences, representing 26.18% of the original rows.

Further investigation showed that 17,781 ZIP-code prefixes were associated with multiple distinct coordinate combinations. This indicates that a single ZIP-code prefix may have several recorded geographic locations.

**Cleaning decision:** Only exact duplicate rows were removed using `drop_duplicates()`. Records with different coordinate combinations were retained, even when they shared the same ZIP-code prefix, city, or state.

The cleaned dataset was stored separately as `geolocation_cleaned`, preserving the original DataFrame for comparison and reproducibility.

**Outcome:** The cleaning process removes redundant exact duplicates while preserving distinct geographic coordinate records.


## 2.4 Data Validation: Checking Negative and Zero Values
Objective - The purpose of this validation check is to identify negative values in the Quantity and Price columns of the cleaned sales dataset.

**Validation Checks**
- Negative Quantities: Identify records where the quantity is less than zero.
- Negative Prices: Identify records where the price is less than zero.
Data Integrity: Investigate flagged records to determine whether they represent data entry errors, product returns, or other valid business transactions.


```python
# Check negative and zero values in important numeric columns

numeric_checks = {
    "order_items": [
        "price",
        "freight_value"
    ],
    "order_payments": [
        "payment_value",
        "payment_installments"
    ],
    "products": [
        "product_weight_g",
        "product_length_cm",
        "product_height_cm",
        "product_width_cm"
    ]
}

for dataset_name, columns in numeric_checks.items():
    df = datasets[dataset_name]

    print(f"\n{'=' * 50}")
    print(f"Dataset: {dataset_name}")
    print(f"{'=' * 50}")

    for column in columns:
        negative_count = (df[column] < 0).sum()
        zero_count = (df[column] == 0).sum()

        print(
            f"{column}: "
            f"Negative values = {negative_count}, "
            f"Zero values = {zero_count}"
        )
```

```python
# Inspect records with zero freight charges

zero_freight = order_items[
    order_items["freight_value"] == 0
]

print("Records with zero freight charges:", len(zero_freight))

display(zero_freight.head(10))
```

```python
print(
    "Unique orders with zero freight:",
    zero_freight["order_id"].nunique()
)

print(
    "Unique products with zero freight:",
    zero_freight["product_id"].nunique()
)
```

```python
# Inspect payments with zero payment values

zero_payments = order_payments[
    order_payments["payment_value"] == 0
]

print("Records with zero payment values:", len(zero_payments))

display(zero_payments)
```

```python
print(
    zero_payments[
        [
            "order_id",
            "payment_sequential",
            "payment_type",
            "payment_installments",
            "payment_value"
        ]
    ].to_string(index=False)
)
```

```python
# Inspect records with zero installments

zero_installments = order_payments[
    order_payments["payment_installments"] == 0
]

print(
    "Records with zero installments:",
    len(zero_installments)
)

display(zero_installments)
```

```python
# Inspect order items with zero or negative prices

suspicious_prices = order_items[
    order_items["price"] <= 0
]

print("Number of records with zero or negative prices:",
      len(suspicious_prices))

display(suspicious_prices.head(10))
```

```python
# Inspect payments with zero or negative values

suspicious_payments = order_payments[
    order_payments["payment_value"] <= 0
]

print("Number of records with zero or negative payment values:",
      len(suspicious_payments))

display(suspicious_payments.head(10))
```

```python
# Inspect products with zero weight

zero_weight = products[
    products["product_weight_g"] == 0
]

print("Products with zero weight:", len(zero_weight))

display(zero_weight)
```

```python
# Scan numeric columns across all datasets for negative values

for name, df in datasets.items():
    numeric_columns = df.select_dtypes(
        include="number"
    ).columns

    for column in numeric_columns:
        count = (df[column] < 0).sum()

        if count > 0:
            print(
                f"{name}.{column}: "
                f"{count} negative values"
            )
```

### Summary of Validation of Numeric Values

Numeric columns were examined for negative and zero values to identify potentially invalid or unusual records.

The initial checks found no negative values in the selected item price, freight, payment, installment, and product measurement columns.

Zero values were identified in the following columns:

* `freight_value`: 383 records
* `payment_value`: 9 records
* `payment_installments`: 2 records
* `product_weight_g`: 4 records
* `Negative values were observed in the latitude and longitude columns of the geolocation dataset. These values were not classified as invalid because geographic coordinates can legitimately be negative. Negative latitude indicates a location south of the Equator, while negative longitude indicates a location west of the Prime Merid`

Item prices had no zero values. Product length, height, and width also had no zero values.

The identified zero values were flagged for further investigation. No automatic replacements or deletions were performed during the initial validation, since the validity of these values depends on their business context.
